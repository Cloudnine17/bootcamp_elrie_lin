````python
import pathlib
#%pip install yfinance
import yfinance as yf

# 1. 確定檔案路徑
REPO_PATH = pathlib.Path.cwd()
raw_data_dir = REPO_PATH / "data" / "raw"
raw_data_dir.mkdir(parents=True, exist_ok=True)
csv_filepath = raw_data_dir / "sp500_daily_history.csv"

# 2. 下載歷史數據 (2020/01/01 ~ 2025/12/31) 並儲存為 CSV
print("Downloading historical S&P 500 data...")
df = yf.download("^GSPC", start="2020-01-01", end="2025-12-31")
df.to_csv(csv_filepath)

print(f"Data successfully saved to: {csv_filepath}")
print(df.head())
````
